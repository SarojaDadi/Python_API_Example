# GitHub Repository Explorer 🔎

A beginner project that uses a **GitHub Personal Access Token** to ask GitHub's API for repository details.

You'll enter your token privately when the notebook runs. **Never put your token in a code cell or commit it to GitHub.**

## 1. Choose a repository

Use a repository's owner and name. This example uses Python's popular `requests` project.

```python
owner = "psf"
repo = "requests"

print(f"Looking up: {owner}/{repo}")
```

## 2. Enter your token securely

When prompted, paste your GitHub Personal Access Token. `getpass` hides what you type, so the token won't appear in the notebook output.

If you don't have a token yet, create one in GitHub under **Settings → Developer settings → Personal access tokens**. For this read-only project, use the smallest permissions needed. For a fine-grained token, grant read access to repository metadata for the repository you want to inspect.

```python
from getpass import getpass

github_token = getpass("GitHub token (input is hidden): ")

if not github_token:
    raise ValueError("A token is required to continue.")

print("Token received. It will not be printed.")
```

## 3. Call the GitHub API

The token is sent in an authorization header over HTTPS. We won't print the request headers or token.

```python
import requests

url = f"https://api.github.com/repos/{owner}/{repo}"
headers = {
    "Authorization": f"Bearer {github_token}",
    "Accept": "application/vnd.github+json",
    "X-GitHub-Api-Version": "2022-11-28",
}

response = requests.get(url, headers=headers, timeout=20)

if response.status_code == 200:
    repo_data = response.json()
    print("Repository details loaded!")
elif response.status_code == 401:
    raise RuntimeError("GitHub rejected the token. Check that it is valid and not expired.")
elif response.status_code == 403:
    raise RuntimeError("Access denied or API rate limit reached. Check token permissions and try later.")
elif response.status_code == 404:
    raise RuntimeError("Repository not found, or your token does not have permission to view it.")
else:
    raise RuntimeError(f"GitHub API returned status {response.status_code}.")
```

## 4. Display the results

```python
print("Name:", repo_data.get("full_name"))
print("Description:", repo_data.get("description") or "No description")
print("Stars:", repo_data.get("stargazers_count"))
print("Open issues:", repo_data.get("open_issues_count"))
print("Main language:", repo_data.get("language"))
print("Repository URL:", repo_data.get("html_url"))
```

## 5. Clean up

Remove the token variable from this notebook's running session when you're finished. This does not revoke the token itself.


```python
github_token = None
headers = None
print("Token variable cleared from this session.")
```
